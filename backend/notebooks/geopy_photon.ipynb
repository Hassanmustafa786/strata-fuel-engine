```python
import time
import pandas as pd
from geopy.geocoders import Photon

df = pd.read_csv("fuel_prices.csv")

# Photon uses OpenStreetMap data without strict rate limits for reasonable batch runs
geolocator = Photon(user_agent="strata_fuel_optimizer")

latitudes = []
longitudes = []

for idx, row in df.iterrows():
  full_address = f"{row['Address']}, {row['City']}, {row['State']}, USA"
  try:
    location = geolocator.geocode(full_address)
    if location:
      latitudes.append(location.latitude)
      longitudes.append(location.longitude)
    else:
      latitudes.append(None)
      longitudes.append(None)
  except Exception as e:
    latitudes.append(None)
    longitudes.append(None)

  if idx % 100 == 0:
    print(f"Processed {idx}/6967 records...")

df["latitude"] = latitudes
df["longitude"] = longitudes

df.to_csv("fuel_prices_geocoded.csv", index=False)
```
